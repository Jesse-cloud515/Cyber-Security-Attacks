## Standardize The Data

```python
# Let's explore the categorical columns in the dataset to understand 

categorical_columns = df_clean.select_dtypes(
    include=['object', 'category']
).columns

print(categorical_columns)
```

```python
# Let's explore the unique values in each categorical column to understand the data better.

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print(df_clean[col].unique())
```

```python
category_columns = [
    'Protocol',
    'Packet Type',
    'Traffic Type',
    'Attack Type',
    'Attack Signature',
    'Action Taken',
    'Severity Level',
    'Network Segment',
    'Log Source'
]
```

```python
# Examine unique values: this columns do not have any inconsistencies, but it is good to check for any unexpected values.

for col in category_columns:
    print(f"\n{col}:")
    print(sorted(df_clean[col].dropna().unique()))
```

```python
# Check for extra spaces in categorical columns

for col in category_columns:

    space_count = (
        df_clean[col]
        .astype('string')
        .ne(
            df_clean[col]
            .astype('string')
            .str.strip()
        )
        .sum()
    )

    print(f"{col}: {space_count} values with extra spaces")
```

```python
# Check for case inconsistencies in categorical columns

for col in category_columns:


    original_count = df_clean[col].nunique()

    normalized_count = (
        df_clean[col]
        .dropna()
        .astype('string')
        .str.strip()
        .str.casefold()
        .nunique()
    )

    print(
        f"{col}: Original = {original_count}, "
        f"Normalized = {normalized_count}"
    )
```

```python
# Investigate inconsistent spacing in other text columns

text_columns = [
    'User Information',
    'Device Information',
    'Payload Data',
    'Geo-location Data',
    'Source IP Address',
    'Destination IP Address',
    'Proxy Information'
]
```

```python
for col in text_columns:

    print(f"\nColumn: {col}")

    inconsistent_spaces = df_clean.loc[
        df_clean[col].str.contains(
            r'\s{2,}',
            na=False
        ),
        col
    ]

    print(inconsistent_spaces.head(10))
```

```python
remaining_spaces = (
    df_clean[text_columns]
    .apply(
        lambda col: col.str.contains(
            r'\s{2,}', na=False
        ).sum()
    )
)

print(remaining_spaces)
```

```python
# Validat IP address formats, to do this we We can use Python's built-in ipaddress module.We can use Python's built-in ipaddress module.

import ipaddress
```

```python
def validate_ip(ip):
    try:
        ipaddress.ip_address(ip)
        return True

    except ValueError:
        return False
```

```python
# Apply the function

for col in ip_columns:

    valid_ips = df_clean[col].dropna().apply(validate_ip)

    invalid_count = (~valid_ips).sum()

    print(f"{col}: {invalid_count} invalid IP addresses")
```

## Identify and investigate outliers

```python
# Identify numerical columns

# Select columns containing numerical data
numeric_columns = df_clean.select_dtypes(include='number').columns

# Display the numerical column names
print(numeric_columns)
```

```python
# Generate descriptive statistics for numerical columns
df_clean[numeric_columns].describe()
```

```python
# Loop through each numerical column
for col in numeric_columns:

    # Calculate the first and third quartiles
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    # Calculate the interquartile range
    IQR = Q3 - Q1

    # Calculate the lower and upper boundaries
    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    # Identify potential outliers
    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    # Display the results
    print(f"\nColumn: {col}")
    print(f"Lower Bound: {lower_bound}")
    print(f"Upper Bound: {upper_bound}")
    print(f"Number of Outliers: {len(outliers)}")
```

```python
# Validate values using domain knowledge

# Check whether source and destination ports are valid
print("Invalid Source Ports:",
      ((df_clean['Source Port'] < 0) |
       (df_clean['Source Port'] > 65535)).sum())

print("Invalid Destination Ports:",
      ((df_clean['Destination Port'] < 0) |
       (df_clean['Destination Port'] > 65535)).sum())

# Check packet lengths
print("Invalid Packet Lengths:",
      (df_clean['Packet Length'] <= 0).sum())

# Check anomaly scores
print("Invalid Anomaly Scores:",
      ((df_clean['Anomaly Scores'] < 0) |
       (df_clean['Anomaly Scores'] > 100)).sum())
```

## Text cleaning and standardisation

```python
# Identify the text columns

# Select columns containing text data
text_columns = df_clean.select_dtypes(include=['object', 'string']).columns

# Display the text column names
print(text_columns.tolist())
```

```python
# Check for unnecessary spaces

# Check each text column for unnecessary spaces
for col in text_columns:

    # Count values with leading or trailing spaces
    outer_spaces = df_clean[col].str.contains(
        r'^\s|\s$',
        na=False
    ).sum()

    # Count values with repeated consecutive spaces
    repeated_spaces = df_clean[col].str.contains(
        r'\s{2,}',
        na=False
    ).sum()

    print(f"\nColumn: {col}")
    print(f"Leading/Trailing Spaces: {outer_spaces}")
    print(f"Repeated Spaces: {repeated_spaces}")
```

```python
# Clean User Information column by removing leading/trailing spaces and replacing multiple consecutive spaces with a single space

# Standardise whitespace in User Information
df_clean['User Information'] = (
    df_clean['User Information']
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)
```

```python
# Check for remaining repeated whitespace
remaining_spaces = df_clean['User Information'].str.contains(
    r'\s{2,}',
    na=False
).sum()

# Check for leading or trailing whitespace
outer_spaces = df_clean['User Information'].str.contains(
    r'^\s|\s$',
    na=False
).sum()

print("Remaining repeated spaces:", remaining_spaces)
print("Remaining leading/trailing spaces:", outer_spaces)
```
