```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the cleaned cybersecurity dataset
df = pd.read_csv(
    'cybersecurity_attacks_cleaned.csv',
    parse_dates=['timestamp', 'event_date']
)
```

# Visualisation and Presentation

We've now covered univariate, categorical, numerical, bivariate and multivariate analysis, including numerical correlation.

The next step is to turn our analysis into clear, professional visualisations that communicate the cybersecurity findings to someone who may not have worked with our dataset.

```python
# Attack Type Distribution

attack_counts = df['attack_type'].value_counts()

plt.figure(figsize=(9, 5))

plt.pie(
    attack_counts.values,
    labels=attack_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Distribution of Recorded Attack Types')
plt.xlabel('Attack Type')
plt.ylabel('Number of Records')
plt.tight_layout()
plt.show()
```

```python
# Severity Distribution

severity_counts = df['severity_level'].value_counts()

plt.figure(figsize=(8, 5))

plt.pie(
    severity_counts.values,
    labels=severity_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)


plt.title('Distribution of Security Severity Levels')
plt.xlabel('Severity Level')
plt.ylabel('Number of Records')
plt.tight_layout()
plt.show()
```

```python
# Attack Type vs Severity

attack_severity_pct = pd.crosstab(
    df['attack_type'],
    df['severity_level'],
    normalize='index'
) * 100

attack_severity_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Severity Distribution by Attack Type')
plt.xlabel('Attack Type')
plt.ylabel('Percentage (%)')
plt.legend(title='Severity Level')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Protocol Distribution

protocol_counts = df['protocol'].value_counts()

plt.figure(figsize=(8,5))

plt.pie(
    protocol_counts.values,
    labels=protocol_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Distribution of Network Protocols')
plt.tight_layout()
plt.show()
```

```python
# Attack Type vs Protocol

protocol_attack_pct = pd.crosstab(
    df['protocol'],
    df['attack_type'],
    normalize='index'
) * 100

protocol_attack_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Attack Type Distribution by Protocol')
plt.xlabel('Protocol')
plt.ylabel('Percentage (%)')
plt.legend(title='Attack Type')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Action Taken Distribution

action_counts = df['action_taken'].value_counts()

plt.figure(figsize=(8, 5))

plt.pie(
    action_counts.values,
    labels=action_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Distribution of Security Actions')
plt.tight_layout()
plt.show()
```

```python
# Anomaly Score Distribution

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x='anomaly_scores',
    bins=30,
    kde=True
)

plt.title('Distribution of Anomaly Scores')
plt.xlabel('Anomaly Score')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Packet Length Distribution

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x='packet_length',
    bins=30,
    kde=True
)

plt.title('Distribution of Network Packet Lengths')
plt.xlabel('Packet Length')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Attack Type vs Anomaly Score

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='attack_type',
    y='anomaly_scores'
)

plt.title('Anomaly Score Distribution by Attack Type')
plt.xlabel('Attack Type')
plt.ylabel('Anomaly Score')
plt.tight_layout()
plt.show()
```

```python
# Events by Hour

hour_counts = df['hour'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
sns.lineplot(
    x=hour_counts.index,
    y=hour_counts.values,
    marker='o'
)

plt.title('Recorded Events by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Number of Records')
plt.xticks(range(24))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Events by Day of Week

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

day_counts = df['day_of_week'].value_counts().reindex(day_order)

plt.pie(
    day_counts.values,
    labels=day_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Recorded Events by Day of Week')
plt.tight_layout()
plt.show()
```

```python
# Events by Day of Week

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

day_counts = df['day_of_week'].value_counts().reindex(day_order)

plt.pie(
    day_counts.values,
    labels=day_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Recorded Events by Day of Week')
plt.tight_layout()
plt.show()
```

```python
# Events by Year

year_counts = df['year'].value_counts().sort_index()

plt.pie(
    year_counts.values,
    labels=year_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white'}
)

plt.title('Recorded Events by Year')
plt.tight_layout()
plt.show()
```

```python
# Attack Type by Year

attack_year_pct = pd.crosstab(
    df['year'],
    df['attack_type'],
    normalize='index'
) * 100

attack_year_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Attack Type Distribution by Year')
plt.xlabel('Year')
plt.ylabel('Percentage (%)')
plt.legend(title='Attack Type')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Numerical Correlation Heatmap

numerical_columns = [
    'source_port',
    'destination_port',
    'packet_length',
    'anomaly_scores'
]

correlation_matrix = df[numerical_columns].corr(
    method='pearson'
)

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title('Correlation Between Numerical Variables')
plt.tight_layout()
plt.show()
```
